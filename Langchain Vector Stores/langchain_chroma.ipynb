2. Create ChromaDB

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma 
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
embeddings=GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001")
```

```python
from langchain_core.documents import Document
documents=[
    Document(
        page_content="Pakistan is located in South Asia",
        metadata={"source":"wiki"}
    ),
    Document(
        page_content="Islamabad is the capital of Pakistan",
        metadata={"source":"wiki"}
    ),
    Document(
        page_content="Karachi is largest city of Pakistan.",
        metadata={"source":"wiki"}
    ),
]
```

```python
vector_store=Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    collection_name="country_info",
    persist_directory="./chroma_db"
)
```

3. Load Existing ChromaDB

```python
from langchain_chroma import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
vector_store=Chroma(
    collection_name="country_info",
    embedding_function=embeddings,
    persist_directory="./chroma_db"
)
print("Database Loaded")
```

4. Similarity Search

```python
query="What is capital?"
results=vector_store.similarity_search(query,k=2)

for result in results:
    print(result.page_content)
    print(result.metadata)
```

5. Similarity Search with Score

```python
results=vector_store.similarity_search_with_score(
    "largest city",
    k=2
)

for doc,score in results:
    print(score)
    print(doc.page_content)
```

6. Convert to Retriever

```python
retriever=vector_store.as_retriever(
    search_kwargs={"k":2}
)
docs=retriever.invoke("largest city")
for doc in docs:
    print(doc.page_content)
```
